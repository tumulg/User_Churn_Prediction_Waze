## Waze's Predict User Churn Project

**Stats and Hypothesis Testing**

$Task:$ To find out whether there is a statistically significant difference in mean amount of rides between the 2 device type users.

*1. Importing and Data Loading*

```python
#Import libraries or packages
import os

#Data Manipulation
import pandas as pd
import numpy as np

#Data Visualization
import matplotlib.pyplot as plt
import seaborn as sns

#Statistics
from scipy import stats
```

```python
#Import the dataset
get_cwd = os.getcwd()
df = pd.read_csv(get_cwd + '/waze_dataset.csv')
```

*2. Data Exploration*

```python
#1. Create `map_dictionary`
map_dictionary = {'Android': 2, 'iPhone': 1}

#2. Create new `device_type` column and map the new column to the dictionary
df['device_type'] = df['device'].map(map_dictionary)
df.head(10)
```

```python
#Calculate the mean of the number of drives for each device type
df.groupby('device_type')['drives'].mean()
```

*3. Hypothesis Testing*

Steps:
1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis

*Step 1:*

Null Hypothesis(H0): There is no difference in avg between drives who use an iPhone and drivers who use an Android.

Alternative Hypothesis(H1): There is a difference in avg between drives who use an iPhone and drivers who use an Android.

*Step 2:*

Significane Level: 5% or 0.05

```python
#Significance Level
sl = 0.05
```

*Step 3: Finding the P-Value* 

```python
#1. Isolate the `drives` column for iPhone users.
iPhone = df[df['device_type'] == 1]['drives']

#2. Isolate the `drives` column for Android users.
Android = df[df['device_type'] == 2]['drives']
```

```python
#3. Perform the t-test
result = stats.ttest_ind(a = iPhone, b = Android, equal_var=False)
p_value = result[1]
print("P-value:", p_value*100)
print("P-value < significance level (5%):", p_value < sl)
```

*Step 4: Reject or Fail to Reject the Null Hypothesis*

Since the P-value is 14.3%, i.e, more than the confidence level, we fail to reject the null hypothesis. You conclude that there is **not** a statistically significant difference in the average number of drives between drivers who use iPhones and drivers who use Androids.

**Conclusion:**

The key business insight is that drivers who use iPhone devices on average have a similar number of drives as those who use Androids.
